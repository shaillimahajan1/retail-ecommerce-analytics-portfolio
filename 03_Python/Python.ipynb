```python
# Raw date format is DD/MM/YYYY HH:MM — required explicit format= in to_datetime(), and notebook state (re-running conversions on already-converted data) caused a silent full data-loss bug that looked identical to a wrong-format bug.
```

```python
# 1. Define the list of your timestamp columns
timestamp_cols = ['order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date', 'order_estimated_delivery_date']

# 2. Loop through and convert each one
for col in timestamp_cols:
    orders[col] = pd.to_datetime(orders[col], errors='coerce', format='%d/%m/%Y %H:%M')

# Verify the changes (it should show 'datetime64[ns]' next to your columns)
print(orders[timestamp_cols].dtypes)

```
